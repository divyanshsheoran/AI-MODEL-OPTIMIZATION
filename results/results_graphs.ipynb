```python
import os
import numpy as np
import matplotlib.pyplot as plt

# -------- Data (post-augmentation results) --------
models = [
    "MobileNetV2\nBaseline",
    "MobileNetV2\nDynamic Quantized",
    "MobileNetV2\nPruned",
    "MobileNetV2\nStudent (KD)",
    "MobileNetV3\nBaseline",
    "MobileNetV3\nDynamic Quantized",
]

# Accuracy (%)
accuracies = [58.26, 58.26, 13.70, 49.78, 89.51, 89.50]

# Model sizes (MB)
sizes = [9.19, 9.19, 18.02, 4.04, 6.25, 4.45]

# Color scheme from previous version
colors = ['#4CAF50', '#2196F3', '#FFC107', '#9C27B0', '#FF5722', '#009688']

# Optional: ensure a results folder exists
os.makedirs("results", exist_ok=True)

x = np.arange(len(models))
bar_width = 0.6

# -------- Accuracy Comparison --------
fig, ax = plt.subplots(figsize=(12, 6))
bars = ax.bar(x, accuracies, bar_width, color=colors)

# Value labels on bars
for b in bars:
    h = b.get_height()
    ax.annotate(f"{h:.2f}%",
                xy=(b.get_x() + b.get_width()/2, h),
                xytext=(0, 3),
                textcoords="offset points",
                ha="center", va="bottom", fontsize=9)

ax.set_ylabel("Accuracy (%)", fontsize=12)
ax.set_title("Model Accuracy Comparison (Post-Augmentation)", fontsize=14)
ax.set_xticks(x)
ax.set_xticklabels(models, rotation=20, ha="right")
ax.set_ylim(0, 100)

plt.tight_layout()
plt.savefig("results/accuracy_comparison_aug.png", dpi=300)
plt.show()

# -------- Model Size Comparison --------
fig, ax = plt.subplots(figsize=(12, 6))
bars = ax.bar(x, sizes, bar_width, color=colors)

for b in bars:
    h = b.get_height()
    ax.annotate(f"{h:.2f} MB",
                xy=(b.get_x() + b.get_width()/2, h),
                xytext=(0, 3),
                textcoords="offset points",
                ha="center", va="bottom", fontsize=9)

ax.set_ylabel("Model Size (MB)", fontsize=12)
ax.set_title("Model Size Comparison (Post-Augmentation)", fontsize=14)
ax.set_xticks(x)
ax.set_xticklabels(models, rotation=20, ha="right")
ax.set_ylim(0, max(sizes) + 2)

plt.tight_layout()
plt.savefig("results/size_comparison_aug.png", dpi=300)
plt.show()
```
